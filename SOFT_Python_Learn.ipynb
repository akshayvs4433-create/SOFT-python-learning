**Task 3 — Resume Generator · Ask for name, education, skills, projects, languages → print a clean professional resume.**

```python
# Task 3 - Resume Generator
name = input("Full Name: ")
education = input("Education: ")
skills = input("Skills (comma separated): ")
projects = input("Projects (comma separated): ")
languages = input("Languages known: ")

print("\n" + "=" * 45)
print(f"{name.upper():^45}")
print("=" * 45)
print("\nEDUCATION")
print("-" * 45)
print(education)
print("\nSKILLS")
print("-" * 45)
for s in skills.split(","):
    print(f"  • {s.strip()}")
print("\nPROJECTS")
print("-" * 45)
for p in projects.split(","):
    print(f"  • {p.strip()}")
print("\nLANGUAGES")
print("-" * 45)
print(languages)
print("=" * 45)
```
